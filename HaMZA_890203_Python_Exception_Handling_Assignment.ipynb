### Q14. Weak Password Rejection

Create a custom exception named `WeakPasswordError`. Ask the user for a password, check both rules, raise the custom exception when a rule fails, and handle it to explain what needs to be corrected. Do not display the actual password in the success message.

```python
class WeakPasswordError(Exception):
    pass

password = input("Create a password: ")
try:
    if len(password) < 8:
        raise WeakPasswordError("Password must contain at least 8 characters.")
    if not any(char.isdigit() for char in password):
        raise WeakPasswordError("Password must contain at least one digit.")
except WeakPasswordError as e:
    print("Password rejected:", e)
else:
    print("Password accepted successfully.")
```

### Q22. Bank Transfer Between Two Accounts

A banking system transfers money from one account to another. The transfer amount must be positive, the sender and receiver must be different accounts, and the sender must have enough funds.

Create custom exceptions named `InvalidTransferError` and `InsufficientFundsError`. Validate all rules and raise the correct exception. Handle the exceptions and ensure that neither account balance changes when any validation fails. Display both balances after the transaction attempt.

```python
class InvalidTransferError(Exception):
    pass

class InsufficientFundsError(Exception):
    pass

account_a = {"name": "Account A", "balance": 10000.0}
account_b = {"name": "Account B", "balance": 4000.0}

try:
    sender = input("Transfer from (A/B): ").strip().upper()
    receiver = input("Transfer to (A/B): ").strip().upper()
    amount = float(input("Enter transfer amount: Rs. "))

    if sender not in ("A", "B") or receiver not in ("A", "B"):
        raise InvalidTransferError("Choose A or B for both accounts.")
    if sender == receiver:
        raise InvalidTransferError("Sender and receiver must be different.")
    if amount <= 0:
        raise InvalidTransferError("Transfer amount must be positive.")

    sender_account = account_a if sender == "A" else account_b
    receiver_account = account_a if receiver == "A" else account_b
    if amount > sender_account["balance"]:
        raise InsufficientFundsError("Sender does not have enough funds.")
except ValueError:
    print("Please enter a valid numeric amount.")
except InvalidTransferError as e:
    print("Transfer rejected:", e)
except InsufficientFundsError as e:
    print("Transfer failed:", e)
else:
    sender_account["balance"] -= amount
    receiver_account["balance"] += amount
    print("Transfer successful.")
print("Account A balance: Rs.", account_a["balance"])
print("Account B balance: Rs.", account_b["balance"])
```

### Q24. Flight Seat Reservation

An airline has a fixed number of seats. A passenger can reserve one or more seats, but the request cannot be zero, negative, or greater than the remaining seats.

Create custom exceptions named `InvalidSeatRequestError` and `SeatsUnavailableError`. Validate user input, raise and handle the appropriate custom exception, and update the remaining-seat count only when the reservation succeeds. Test a valid booking and at least two invalid booking attempts.

```python
class InvalidSeatRequestError(Exception):
    pass

class SeatsUnavailableError(Exception):
    pass

remaining_seats = 5

def reserve_seats(request):
    global remaining_seats
    if request <= 0:
        raise InvalidSeatRequestError("Seat request must be at least one.")
    if request > remaining_seats:
        raise SeatsUnavailableError("Not enough seats remain.")
    remaining_seats -= request
    print(f"Successfully reserved {request} seat(s).")

# Demonstrate one valid and two invalid booking attempts.
test_requests = [2, 0, 10]
for request in test_requests:
    try:
        print("Trying to reserve:", request)
        reserve_seats(request)
    except InvalidSeatRequestError as e:
        print("Invalid request:", e)
    except SeatsUnavailableError as e:
        print("Reservation failed:", e)
print("Seats remaining:", remaining_seats)
```

### Q28. Bulk Employee Payroll Processing

A company stores several employee records in a list. Each record contains an employee name, hours worked, and hourly rate. One invalid record should not stop the payroll calculation for all other employees.

Create a custom exception named `InvalidPayrollRecordError`. Validate each employee record, raise this exception when required data is missing or values are invalid, and handle the error inside the record-processing loop. Calculate payroll for valid employees and report which records were skipped and why.

```python
class InvalidPayrollRecordError(Exception):
    pass

employees = [
    {"name": "Ali", "hours": 40, "rate": 500},
    {"name": "Sara", "hours": 35, "rate": 600},
    {"name": "Ahmed", "hours": -5, "rate": 450},
    {"name": "Hina", "hours": 42},  # Missing hourly rate
    {"name": "Usman", "hours": 45, "rate": 550}
]

total_payroll = 0
for employee in employees:
    try:
        name = employee["name"]
        hours = employee["hours"]
        rate = employee["rate"]
        if not name:
            raise InvalidPayrollRecordError("Employee name is missing.")
        if not isinstance(hours, (int, float)) or not isinstance(rate, (int, float)):
            raise InvalidPayrollRecordError("Hours and rate must be numeric.")
        if hours < 0 or hours > 80:
            raise InvalidPayrollRecordError("Hours must be between 0 and 80.")
        if rate <= 0:
            raise InvalidPayrollRecordError("Hourly rate must be greater than zero.")
    except KeyError as e:
        print("Skipped record", employee, "- missing field:", e)
    except InvalidPayrollRecordError as e:
        print("Skipped", employee.get("name", "Unknown"), "-", e)
    else:
        pay = hours * rate
        total_payroll += pay
        print(employee["name"], "pay: Rs.", pay)

print("Total payroll for valid employees: Rs.", total_payroll)
```

Use `try`, `except`, `else`, and `finally` meaningfully: report success only in `else`, handle expected failures in `except`, and display an audit/attempt-finished message in `finally`.
```python
# class ProductNotFoundError(Exception):
    pass

class InvalidQuantityError(Exception):
    pass

class InsufficientStockError(Exception):
    pass

inventory = {"P101": 10, "P102": 5}
product_id = input("Enter product ID: ").strip().upper()
try:
    quantity = int(input("Enter purchase quantity: "))
    if product_id not in inventory:
        raise ProductNotFoundError("Product ID does not exist.")
    if quantity <= 0:
        raise InvalidQuantityError("Quantity must be a positive integer.")
    if quantity > inventory[product_id]:
        raise InsufficientStockError("Not enough stock for this purchase.")
except ValueError:
    print("Please enter a valid whole-number quantity.")
except ProductNotFoundError as e:
    print("Purchase failed:", e)
except InvalidQuantityError as e:
    print("Purchase failed:", e)
except InsufficientStockError as e:
    print("Purchase failed:", e)
else:
    inventory[product_id] -= quantity
    print("Purchase successful.")
    print("Remaining stock:", inventory[product_id])
finally:
    print("Inventory purchase attempt finished.")
print("Current inventory:", inventory)
```

### Q30. Mini Banking System with a Custom Exception Hierarchy

Design a menu-based mini banking program that supports deposit, withdrawal, and balance viewing. Deposits and withdrawals must be positive. A withdrawal cannot exceed the available balance. Invalid menu choices must also be handled.

Create a base custom exception named `BankingError`, then create `InvalidAmountError` and `InsufficientBalanceError` as child classes of `BankingError`. Raise the appropriate exception from the relevant operation and handle the base class or specific child classes safely. Keep the program running after an invalid transaction, and show the final account balance when the user chooses to exit.

```python
class BankingError(Exception):
    pass

class InvalidAmountError(BankingError):
    pass

class InsufficientBalanceError(BankingError):
    pass

balance = 5000.0

while True:
    print("\n--- MINI BANKING SYSTEM ---")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. View balance")
    print("4. Exit")
    choice = input("Choose an option (1-4): ").strip()

    if choice == "4":
        break

    try:
        if choice == "1":
            amount = float(input("Enter deposit amount: Rs. "))
            if amount <= 0:
                raise InvalidAmountError("Deposit must be positive.")
            balance += amount
            print("Deposit successful.")
        elif choice == "2":
            amount = float(input("Enter withdrawal amount: Rs. "))
            if amount <= 0:
                raise InvalidAmountError("Withdrawal must be positive.")
            if amount > balance:
                raise InsufficientBalanceError("Insufficient account balance.")
            balance -= amount
            print("Withdrawal successful.")
        elif choice == "3":
            print("Current balance: Rs.", balance)
        else:
            raise BankingError("Invalid menu choice. Select 1, 2, 3, or 4.")
    except ValueError:
        print("Please enter a valid numeric amount.")
    except BankingError as e:
        print("Transaction error:", e)

print("Final account balance: Rs.", balance)
print("Thank you for using the mini banking system.")
```
